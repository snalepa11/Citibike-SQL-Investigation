```python
# Convert ride_date to datetime and extract year
df['ride_date'] = pd.to_datetime(df['ride_date'])
df['year'] = df['ride_date'].dt.year

# Get unique years
years = df['year'].unique()

# Create separate graph for each year
for year in sorted(years):
    df_year = df[df['year'] == year]
    
    plt.figure(figsize=(12, 4))
    plt.bar(df_year['ride_date'].dt.strftime('%Y-%m-%d'), df_year['num_rides'])
    plt.title(f'Number of Rides in {year}')
    plt.xlabel('Date')
    plt.ylabel('Number of Rides')
    plt.xticks(rotation=90)
    plt.tight_layout()
    plt.show()
```

We can see a general along the seasons with a steady incline of ridership over the years. 

```python
# Compare ridership across days of the week

# Define the order of days for proper sorting
day_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']

# Calculate mean ridership by day of week
mean_by_day = df.groupby('day_of_week')['num_rides'].mean().reindex(day_order)

# Create bar chart of mean ridership
plt.figure(figsize=(10, 5))
plt.bar(day_order, mean_by_day)
plt.title('Mean Ridership by Day of Week')
plt.xlabel('Day of Week')
plt.ylabel('Mean Number of Rides')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Create box plots
plt.figure(figsize=(12, 6))
df['day_of_week'] = pd.Categorical(df['day_of_week'], categories=day_order, ordered=True)
df_sorted = df.sort_values('day_of_week')
df_sorted.boxplot(column='num_rides', by='day_of_week', figsize=(12, 6))
plt.title('Distribution of Ridership by Day of Week')
plt.suptitle('')  # Remove default title
plt.xlabel('Day of Week')
plt.ylabel('Number of Rides')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Print statistics
print("Mean ridership by day of week:")
print(mean_by_day.round(0))

#E5 : Based on the graphs below, I would think it's safe to say that bike users are primarily commuters as we can see an uptick of usage during week days and a dip over the weekend. 
```
